```python
import yfinance as yf
import pandas as pd

# Changes from original list:
# FANG  → NFG  (National Fuel Gas)     | FANG IPO Oct 2012
# PSX   → FLR  (Fluor Corporation)     | PSX spun off May 2012
# BKR   → FTI  (TechnipFMC)            | BKR data only from 2017

energy_tickers = [
    "XOM",  # ExxonMobil
    "CVX",  # Chevron
    "COP",  # ConocoPhillips
    "SLB",  # Schlumberger
    "EOG",  # EOG Resources
    "FLR",  # Fluor Corporation        ← was PSX
    "VLO",  # Valero Energy
    "OXY",  # Occidental Petroleum
    "HAL",  # Halliburton
    "FTI",  # TechnipFMC               ← was BKR
    "DVN",  # Devon Energy
    "NFG",  # National Fuel Gas        ← was FANG
    "APA",  # APA Corporation
    "OKE",  # ONEOK
    "WMB",  # Williams Companies
]

start_date = "2005-01-01"
end_date   = "2024-12-31"

data = yf.download(energy_tickers, start=start_date, end=end_date, auto_adjust=True)

close_price = data["Close"]

close_price.to_csv("Energy_close_price.csv")

print(close_price.head())
print(f"\n✅ Downloaded {close_price.shape[1]} stocks × {close_price.shape[0]} trading days")
print(f"\nMissing values per stock:\n{close_price.isna().sum()}")
```

```python
"""
CSV Verification Script
========================
Verifies all 4 sector CSV files before preprocessing.
Checks: shape, date range, null values, duplicate dates, data types.

Run from the folder where your 4 CSV files are located.
"""

import pandas as pd
import numpy as np

FILES = {
    "Energy"    : "Energy_close_price.csv",
    "IT"        : "IT_close_price.csv",
    "Financial" : "Financial_close_price.csv",
    "Healthcare": "Healthcare_close_price.csv",
}

EXPECTED_STOCKS    = 15
EXPECTED_START     = "2005-01-01"
EXPECTED_END       = "2024-12-31"
MIN_TRADING_DAYS   = 4900   # ~20 years × 252 trading days

print("=" * 60)
print(" CSV Verification Report")
print("=" * 60)

all_passed = True

for sector, filename in FILES.items():
    print(f"\n{'─'*60}")
    print(f" Sector : {sector}")
    print(f" File   : {filename}")
    print(f"{'─'*60}")

    # ── Load ──────────────────────────────────────────────
    try:
        df = pd.read_csv(filename, index_col=0, parse_dates=True)
    except FileNotFoundError:
        print(f"  ❌ FILE NOT FOUND: {filename}")
        all_passed = False
        continue

    # ── Shape ─────────────────────────────────────────────
    rows, cols = df.shape
    shape_ok = (cols == EXPECTED_STOCKS) and (rows >= MIN_TRADING_DAYS)
    print(f"  Shape            : {rows} rows × {cols} columns  {'✅' if shape_ok else '❌ (unexpected)'}")

    # ── Date range ────────────────────────────────────────
    actual_start = df.index.min().date()
    actual_end   = df.index.max().date()
    date_ok = (str(actual_start) <= EXPECTED_START[:10] or str(actual_start)[:7] == "2005-01") \
              and str(actual_end)[:4] == "2024"
    print(f"  Date range       : {actual_start} → {actual_end}  {'✅' if date_ok else '⚠️ check dates'}")

    # ── Null values ───────────────────────────────────────
    null_counts = df.isna().sum()
    total_nulls = null_counts.sum()
    null_ok = total_nulls == 0
    print(f"  Total nulls      : {total_nulls}  {'✅' if null_ok else '❌'}")
    if not null_ok:
        print(f"  Nulls per ticker :")
        print(null_counts[null_counts > 0].to_string(header=False))

    # ── Duplicate dates ───────────────────────────────────
    dup_dates = df.index.duplicated().sum()
    dup_ok = dup_dates == 0
    print(f"  Duplicate dates  : {dup_dates}  {'✅' if dup_ok else '❌'}")

    # ── Data type ─────────────────────────────────────────
    non_numeric = [c for c in df.columns if not pd.api.types.is_numeric_dtype(df[c])]
    dtype_ok = len(non_numeric) == 0
    print(f"  All numeric      : {'Yes ✅' if dtype_ok else f'No ❌ — {non_numeric}'}")

    # ── Negative or zero prices ───────────────────────────
    bad_prices = (df <= 0).sum().sum()
    price_ok = bad_prices == 0
    print(f"  Zero/neg prices  : {bad_prices}  {'✅' if price_ok else '❌'}")

    # ── Tickers ───────────────────────────────────────────
    print(f"  Tickers          : {list(df.columns)}")

    # ── Overall ───────────────────────────────────────────
    sector_ok = shape_ok and null_ok and dup_ok and dtype_ok and price_ok
    print(f"\n  {'✅ PASSED' if sector_ok else '❌ ISSUES FOUND — fix before preprocessing'}")
    if not sector_ok:
        all_passed = False

print(f"\n{'='*60}")
print(f" Overall result: {'✅ All 4 files passed — ready for preprocessing!' if all_passed else '❌ Fix the issues above before proceeding.'}")
print(f"{'='*60}")
```
